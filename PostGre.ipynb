```python
import pandas as pd
from sqlalchemy import create_engine
from getpass import getpass
```

```python
senha = getpass("Senha:")
endpoint = getpass("Endpoint:")
```

```python
url = f"postgresql+psycopg://postgres:{senha}@{endpoint}:5432/postgres"
print(url)
```

```python
engine = create_engine(url)
```

```python
with engine.begin() as conexao:
  print("Conectado ao banco de dados")
```

```python
tabela = {
    "sensor" : ["S01","S02","S03"],
    "setor" : ["Montagem","Usinagem","Pintura"],
    "leituras" : [72.5,81.3,32.6]

} #Criando um ".csv"

dados = pd.DataFrame(tabela)
dados.head()
```

```python
dados.to_sql("sensores",
             engine,
             if_exists="replace",
             index=False)
```

```python
teste = pd.read_sql("SELECT * FROM sensores",engine)
teste.head()
```
